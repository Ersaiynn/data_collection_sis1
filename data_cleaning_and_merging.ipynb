# Data Cleaning & Merging (Data Engineer)

Бұл дәптер екі бастапқы CSV файлын (`weather_raw.csv` және `tourism_raw.csv`) тазалау және біріктіру процесін орындайды.

### Қадамдар:
1. **Data Cleaning:** Деректерді тексеру (бос мәндер, дубликаттар) және типтерді дұрыстау.
2. **Merging:** Екі датасетті `City` бағаны бойынша біріктіру.
3. **Export:** Итоговый датасетті `weather_tourism_cleaned_v2.csv` форматында сақтау.

```python
import pandas as pd


weather_df = pd.read_csv('weather_raw.csv')
tourism_df = pd.read_csv('tourism_raw.csv')


merged_df = pd.merge(weather_df, tourism_df, on='City', how='inner')

merged_df.columns = merged_df.columns.str.lower()


print("Пропущенные значения:")
print(merged_df.isnull().sum())
print("\nКоличество дубликатов:", merged_df.duplicated().sum())


merged_df.to_csv('weather_tourism_cleaned_v2.csv', index=False)


merged_df
```
